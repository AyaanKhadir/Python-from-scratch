Q19. Loop with Conditions
Create a list of marks.
Use a for loop to print each mark along with "Pass" or "Fail".
Example:
80 Pass
45 Fail


```python
students = [
    {"Name" : "Rouman" , "Marks": 90},
    {"Name" : "Hari" , "Marks": 60},
    {"Name" : "Rahul" , "Marks": 70}
]

for student in students:
  if student["marks"] >=70:
    print(student["name"], "Passed")
  else:
    print(student["name"], "Failed")
```
